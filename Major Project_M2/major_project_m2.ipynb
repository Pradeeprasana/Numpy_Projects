```python
import numpy as np

def arithmetic_operations(arr1, arr2):
    print("\nBasic Arithmetic Operations:")
    print("Addition:\n", np.add(arr1, arr2))
    print("Subtraction:\n", np.subtract(arr1, arr2))
    print("Multiplication:\n", np.multiply(arr1, arr2))
    print("Division:\n", np.divide(arr1, arr2))
    print("Power (arr1 ** arr2):\n", np.power(arr1, arr2))

def statistical_operations(arr):
    print("\nStatistical Operations:")
    print("Mean:", np.mean(arr))
    print("Median:", np.median(arr))
    print("Standard Deviation:", np.std(arr))
    print("Variance:", np.var(arr))
    print("Minimum:", np.min(arr))
    print("Maximum:", np.max(arr))
    print("Sum:", np.sum(arr))
    print("Product:", np.prod(arr))
    print("Shape:", arr.shape)
    print("Size:", arr.size)
    print("Data Type:", arr.dtype)

def array_manipulation(arr):
    print("\nArray Manipulation:")

    try:
        new_shape = (int(input("Enter new rows: ")), int(input("Enter new columns: ")))
        reshaped = arr.reshape(new_shape)
        print("Reshaped Array:\n", reshaped)
    except:
        print("Cannot reshape array with given dimensions.")

    print("\nSorted Array:", np.sort(arr))
    print("Reverse Sorted Array:", np.sort(arr)[::-1])
    print("Unique Values:", np.unique(arr))
    print("First Element:", arr[0])
    print("Last Element:", arr[-1])

    mean_value = np.mean(arr)
    print("Filtering values greater than mean (", mean_value, "):")
    print(arr[arr > mean_value])


print("NumPy Mathematical Calculator")

arr1 = np.array(list(map(float, input("\nEnter elements of first array separated by space: ").split())))
arr2 = np.array(list(map(float, input("Enter elements of second array separated by space: ").split())))

while True:
    print("""
==========================================
             Main Program
==========================================
==========================================
🔧 Select an Operation
1. Basic Arithmetic Operations
2. Statistical Operations (on first array)
3. Array Manipulation (on first array)
4. Exit
==========================================
""")
    
    choice = input("Enter your choice (1-4): ")

    if choice == '1':
        arithmetic_operations(arr1, arr2)
    elif choice == '2':
        statistical_operations(arr1)
    elif choice == '3':
        array_manipulation(arr1)
    elif choice == '4':
        print("Exiting")
        break
    else:
        print("Invalid choice. Please try again.")
```
